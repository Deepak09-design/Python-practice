```python
# String Immutable means you cannot modify the existing string after it is created 
word = "PYTHON"
word[0] = "J"
print (word)
```

## Mutable - add, update , remove , insert , extend, sort

```python
# update method
fruits = ["apple", "banana", "orange"]
fruits[0] = "mango"
print(fruits)
#python replace "apple" with "mango" its mutable 
```

```python
#append : by default adds a new element to end of the list 
fruits = ["apple", "banana", "orange"]
fruits.append("mango") # fucntion methos use parentheses ()
print(fruits)
```

```python
#append with multiple elements example annd 
fruits = ["apple", "banana", "orange"]
fruits.append(["mango", "cherry"]) # python adds the entire list consider as one single element 
print(fruits)
```

```python
#extend - Adds as a individual element also add as last element 
fruits = ["apple", "banana", "orange"]
fruits.extend(["mango", "cherry"]) 
print(fruits)
```

```python
#insert method add element into speecific position
fruits = ["apple", "banana", "orange"]
fruits.insert(1, "mango")
print(fruits)
```

```python
#example-2
#insert method add element into speecific position
fruits = ["apple", "banana", "orange"]
fruits.insert(1, "grape")
print(fruits)
```

```python
##remove() remove element by value 
## remove method not support element by index
fruits = ["apple", "banana", "orange"]
fruits.remove("banana")
print(fruits)
```

```python
# we can use del method to delete specific element by calling index 
#del mathod not support by element value 
fruits = ["apple", "banana", "orange"]
del fruits[1]
print(fruits)
```

```python
# pop method Remove last element 
fruits = ["apple", "banana", "orange"]
fruits.pop()
print(fruits)
```

```python
# pop method Remove using index specific element we can remove  
fruits = ["apple", "banana", "orange"]
fruits.pop(1)
print(fruits)
```

## Dictionary method
## dictionary dtores the data as key-value pairs

```python
student = {
    "name": "deepak",
    "cousre": "FDE",
    "age" : 21
}
print(student)
```

```python
#specific key calling 
student = {
    "name": "deepak",
    "cousre": "FDE",
    "age" : 21
}
print(student["name"])
```

```python
#specific key calling 
# if i declare duplicate values allowed laste value
student = {
    "name": "deepak",
    "cousre": "FDE",
    "age" : 21,
    "age" : 22
}
print(student)
```

```python
## update 
student = {
    "name": "deepak",
    "cousre": "FDE",
    "age" : 21
}
student["age"] = 22
print(student)
```

```python
#adding new record
student = {
    "name": "deepak",
    "cousre": "FDE",
    "age" : 20
}
student["city"] = "hyd"
print(student)
```

```python
#adding new record
student = {
    "name": "deepak",
    "cousre": "FDE",
    "age" : 21
}
student["student_name"] = student.pop("name")
print(student)
```
