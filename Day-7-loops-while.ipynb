### While loop

### while loop is used to repeat block of code as long as given conditon true
### while loop can use index (i) to access list of elements
### for loop is used to iterate over the sequence of elements (list, tuple, dictionery )

```python
#infinity loop
while True:
    print("hello")
```

```python
#infinity loop print 1 
i=1
while True:
    print(i)
```

```python
#infinity loop print sequence order 0,1,2    .......
i=0
while True:
    print(i)
    i=i+1 #incremental operator 
```

```python
i = 1
while i in range (1,5):
    password = input("Enter your password :")
    if password == "aws":
        print("allow to login")
    else:
        print("not allow")
    i = i+1
    
    
```

```python
## for loop exmaple 
for i in range(1,4):
    question = input("ASK to AI :")
    if question.lower() == "exit":
        print("chatbot stopped")
        break
    print("AI is responding")
#i can able to ask howmany questions? 3 questions 
```

```python
for i in range (1,6):
    if i == 2:
        continue
    if i == 4:
        break
    print(i)
    # continue --> skip 2
    # break -----> stop the loop
```
