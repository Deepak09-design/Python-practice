```python
def get_age():
    age = int(input("Enter the age :"))
    return (age)

user_age = get_age()
if user_age >= 18:
    print("major")
else:
    print("minor")
```

```python
def caliculate_final():
    price = 1000
    discount = 10
    final_price = price -(price * discount / 100)
    return final_price
caliculate_final()
```

```python
# without function
### first person salary ###
salary1 = 60000
monthly1 = salary1 / 12 
print(monthly1)

### second person salary ###
salary2 = 120000
monthly2 = salary2 / 12 
print(monthly2)

### third person salary ###
salary3 = 240000
monthly3 = salary3 / 12 
print(monthly3)
```

```python
# with function 
def calculate_monthly_salary(salary):
    monthly = salary / 12
    return monthly 
result1=calculate_monthly_salary(60000)
print(result1)
result2=calculate_monthly_salary(120000)
print(result2)
result3=calculate_monthly_salary(240000)
print(result3)
```
