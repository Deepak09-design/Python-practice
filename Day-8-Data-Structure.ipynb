### 1.1.String Indexing
### indexing :we can call individual letter from word

                                            ## P   Y   T   H   O    N


### positive index start from 0(left) ---->    0   1  2  3  4    5

### Negative index start from -1 (right) <--- -5   -4  -3   -2   -1

```python
#positive index
word = "PYTHON"

print(word[0])
print(word[3])
print (word[5])
```

```python
## negtive index
word = "PYTHON"
print(word[-6])
print(word[-4])
```

```python
word = "PYTHON"
print("last character:", word[-6])
```

# String Slicing
## slicing means extracting a portion of srting
## start index where slicing begins, included
## stop index where slicing ends, excluded
## step number of positions to shift
## Positive Step : start should be less than end (left to right) incremental slicing
## negative Step : start shoulod be grated than stop (right to left) Decremental slicing
## Ex: [start :stop : step]
## P Y T H O N
## 0 1 2 3 4 5
## -6 -5 -4 -3 -2 -1

```python
word = "PYTHON"
print(word[3:5])
#start is 3
#stop is 4
```

```python
word = "PYTHON"
print(word[2:5])
#start is 3
#stop is 4
## python slicing not give any erro if stop range is out of the range and als start point larger than end point 
```

```python
#omitting the start index
word = "PYTHON"
print(word[:3])
print(word[:4])
print(word[:10])
```

```python
#omitting the stop index
word = "PYTHON"
print(word[3:])
print(word[1:])
print(word[:])
```

```python
#Negative slicing process
word = "PYTHON"
print(word[-4:3])
```

```python
word = "PYTHON"
print(word[-4:2])
```

```python
word = "PYTHON"
print(word[-4:-5:-1])
```

```python
word = "PYTHON"
print(word[-4:-6:-1])
```
